# **Visualización del redimiento de los usuarios lo largo del tiempo**


```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="darkgrid")
sns.barplot(
    data=df_usuario,
    x="Año",
    y="Total",
    hue="Año",
    legend=False,
    palette="viridis"
)
plt.title("Total de usuarios atendidos por año")
plt.xlabel("Año")
plt.ylabel("Total de usuarios atendidos")
plt.tight_layout()
plt.show()
```

**¿Cómo ha evolucionado la atención al cliente desde 2019 al 2026?**

La atención al cliente ha evolucionado de forma exponencial. Lo que significa que, a medida que los años han ido avanzando, la cantidad de usuarios atendidos también estuvo ascendiendo. Únicamente en 2024 hubo una ligera variación, pero después sigue representando el mismo patrón.

# **Visualización de los canales más utilizados**

```python
canales = [
    "Presencial",
    "Correo_electrónico",
    "Vía_telefónica",
    "Chat",
    "Whatsapp",
    "Atención_virtual",
    "Redes_sociales",
    "IVR",
    "Consultas_página_web",
    "ChatBot_Whatsapp",
    "ChatBot_web"
]

ranking_canales = (
    df_usuario[canales]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

ranking_canales.columns = ["Canal", "Usuarios"]
plt.figure(figsize=(12,7))

sns.barplot(
    data=ranking_canales,
    x="Usuarios",
    y="Canal",
    hue="Canal",
    palette="viridis",
    legend=False
)

plt.title(
    "Canales de atención más utilizados",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Cantidad de usuarios")
plt.ylabel("Canal")

plt.tight_layout()
plt.show()
```

**Rendimiento de las vías de comunicación: ¿cuáles canales poseen un mayor uso y crecimiento a lo largo del tiempo?**

El canal más utilizado es el ChatBot Web. Este medio de atención funciona gracias a su habilidad para responder de forma inmediata y personalizar el servicio, aprovechando los algoritmos del procesamiento del lenguaje natural (NLP) para interpretar y humanizar las conversaciones con el cliente.

El segundo canal más utilizado es la vía telefónica. Este medio ofrece una conexión más personal y una comunicación instantánea con las preocupaciones del cliente. Además, genera buenas primeras impresiones para retener usuarios.



# **Insights y recomendaciones**
